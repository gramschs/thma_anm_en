## How do we store measured values in variables?

Imagine a sensor on a test vehicle measures the current speed. We want to
store this value along with further information about the measurement in our
program.

```
speed = 27.8
num_measurements = 50
sensor_name = 'Sensor_A'
```

We call `speed`, `num_measurements` and `sensor_name` **variables**. A
variable is like a labeled drawer: we give it a name and put a value inside.
Putting a value inside is called **assignment**, and we use the equals sign
`=` for it. Technically, the drawer is a small area in the computer's memory.

Next we print the content of the variables to the monitor or display. For
this we use Python's built-in `print()` function.

```
print(speed)
print(num_measurements)
print(sensor_name)
```

We recognize functions by the parentheses after the function name. Into
these parentheses goes the so-called **argument**. An argument is, so to
speak, the input passed to the function so that Python knows which value to
display on the screen.

Python distinguishes the values in a variable by their **data type**. In
this section we introduce three simple data types:

* **Integer**: a whole number, for example the number of measurements.
* **Float**: a decimal number, for example a speed in km/h. The decimal
  separator in Python is always a point, never a comma.
* **String**: a sequence of characters, for example the name of a sensor.
  Strings are enclosed in single or double quotation marks.

With the `type()` function we display the data type of a variable.

```
print(type(speed))
print(type(num_measurements))
print(type(sensor_name))
```

Variables can also be changed later. We often write code like the following
example to increase a variable by a value.

```
# TODO: ???   increase num_measurements by 1
print(num_measurements)
```

*Doesn't that contradict mathematics?* In mathematics we would be allowed to
subtract `num_measurements` from both sides of the equation and would arrive
at `0 = 1`, which is obviously false. In Python, however, the
equals sign `=` is not a mathematical equation but an **assignment**. Python
first computes the value on the right-hand side and then stores the result
in the variable on the left-hand side.

For floats we can additionally specify the number of decimal places. For

```
pi = 3.141592653589793
print(f'Pi to two decimal places: {pi:.2f}')
```

`!=` for "is not equal to". The result of a comparison is always a **truth

```
speed_limit_ms = 33.3   # corresponds to 120 km/h

# TODO: ???   print speed_ms > speed_limit_ms and speed_ms == speed_limit_ms
```

## How does our program control the flow?

So far, every line of our program was executed exactly once. Often, however,
we want to run different code depending on a condition, for example depending
on whether a speed limit is being observed.

```
speed_ms = 35.0
speed_limit_ms = 33.3

if speed_ms > speed_limit_ms:
    pass   # TODO: ???   print 'Speed too high!'
elif speed_ms == speed_limit_ms:
    pass   # TODO: ???   print 'Speed exactly at the limit.'
else:
    pass   # TODO: ???   print 'Speed within the permitted range.'
```

We call this structure a **branch**. The line with `if` is called the
**header** and is terminated with a colon `:`. All statements that belong to
this condition are indented by the same number of spaces. This **indentation**
shows Python which code belongs together. With `elif` (short for "else if")
we check a further condition, with `else` we catch all remaining cases.

Often we need to repeat a statement several times, for example to simulate
several time steps of a measurement. For this we use the **for loop**
together with the `range()` function.

```
for time_step in range(5):
    pass   # TODO: ???   print 'Time step {time_step}: measurement is being carried out.'
```

The header consists of the fixed keywords `for` and `in`. Between them is
`time_step`, the **loop variable**. On each pass it takes the next value from
`range(5)`, that is, 0, 1, 2, 3 and 4 in turn, because `range(5)` generates
five numbers starting at 0. The indented lines below form the **loop body**
and are executed on each pass.

Within the loop body we may also use a branch, for example to check for each
time step whether a limit value is exceeded.

```
for time_step in range(5):
    speed_ms = time_step * 12.0
    if speed_ms > speed_limit_ms:
        print(f'Time step {time_step}: {speed_ms:.1f} m/s, too fast!')
    else:
        print(f'Time step {time_step}: {speed_ms:.1f} m/s, within the limit.')
```

The loop runs through all five time steps. On each pass we compute the
current speed and decide with the `if` check which message is printed.
