## What is a Jupyter Notebook?

A Jupyter notebook is an interactive document that combines text, Python code
and outputs in a single file. We recognize notebook files by the `.ipynb`
extension. A notebook consists of individual **cells**. There are two kinds:

* **Code cells** contain Python code. When we run a code cell, Python
  evaluates the code and shows the result directly below it.
* **Text cells** contain explanatory text. They are written in the
  **Markdown** language and are displayed as formatted text when run.

These lecture notes are themselves built from notebooks: the running text
sits in text cells, and the gray-shaded code blocks are executable code
cells.

## Running a cell

We click the cell and then press **Shift+Enter**. Alternatively, there is a
button with a triangle (`Run`) in the toolbar. The result appears immediately
below the cell.

````
3 * (7 - 10) + 5
````

If the last line of a code cell evaluates to a value, the notebook displays
that value automatically. If we want to see several values or intermediate
results, we use the `print()` function.

````
print(3 / 4)
print(2 ** 10)
````

To the left of an executed code cell there is a number in square brackets,
for example `[1]`. This is the **execution counter**. It shows in which
position this cell was run.

Shift+Enter runs the cell and moves to the next one. If we want to stay in
the same cell after running it, we press **Ctrl+Enter**.

With the `+` button in the toolbar we insert a new cell. We delete a selected
cell with the trash-can icon or, when the cell has a blue border, by pressing
the `d` key twice.

## Order matters

A notebook does **not** run cells from top to bottom by itself. It runs them
in the order in which we click and start them. The execution counter `[1]`,
`[2]`, `[3]` shows this order.

This matters as soon as one cell builds on another. In the following example,
the first cell creates the variable `mass`, and the second one computes with
it.

````
mass = 1200
````

````
print(mass * 2)
````

If we run the second cell without having run the first one beforehand, Python
does not yet know the variable `mass` and reports a `NameError`.

If the state of the notebook has become muddled, a restart helps: from the
`Kernel` menu we choose `Restart Kernel and Run All Cells`. This makes Python
forget all variables and run the whole notebook afresh from top to bottom.

## When an error message appears

Error messages are normal and part of programming. The **last line** of the
message names the error type and a short description, for example:

```code
NameError: name 'mass' is not defined
```

This message tells us: Python does not know the variable `mass`. Usually this
is due to a typo in the name or to the cell with the assignment not having
been run yet.
