# Python Cheat Sheet

This page summarizes the Python building blocks from Part 1 at a glance. It
is intended as a reference aid for the exercises and the exam.

## Variables and data types

```
speed = 27.8          # float (decimal number, decimal point)
count = 50            # integer (whole number)
name = 'Sensor_A'     # string (text in quotation marks)

type(speed)           # shows the data type: <class 'float'>
```

An **assignment** with `=` first computes the right-hand side and stores the
result in the variable on the left. `count = count + 1` increases `count` by
1.

## Output with print() and f-strings

```
print(speed)                      # simple output
print(f'Value: {speed} km/h')     # f-string: variable in { }
print(f'Value: {speed:.2f} km/h') # rounded to 2 decimal places
```

## Operators

| Kind | Operators |
| --- | --- |
| Arithmetic | `+`  `-`  `*`  `/`  `**` (power) |
| Comparison | `>`  `<`  `>=`  `<=`  `==` (equal)  `!=` (not equal) |
| Logic | `and`  `or`  `not` |

A comparison always yields `True` or `False`. `=` is the assignment, `==` is
the comparison.

## Branch: if / elif / else

```
if speed > 33.3:
    print('too fast')
elif speed == 33.3:
    print('exactly at the limit')
else:
    print('within the limit')
```

The header ends with `:`, the associated block is indented.

## Repetition: for loop

```
for time_step in range(5):           # 0, 1, 2, 3, 4
    print(time_step)

for value in measurement_series:     # directly over a list
    print(value)
```

`range(5)` generates 0 to 4. `range(0, 101, 20)` generates 0, 20, 40, 60, 80,
100.

## List

```
measurement_series = [80, 95, 120, 60, 110]

measurement_series[0]          # first element: 80
measurement_series[-1]         # last element: 110
len(measurement_series)        # number of elements: 5
measurement_series.append(75)  # append a value at the end
```

## Dictionary

```
sensor = {
    'name': 'Sensor_A',
    'measuring_range_max': 200.0,
}

sensor['name']                       # access via the key: 'Sensor_A'
sensor['measuring_range_max'] = 250.0   # change a value
sensor['calibrated'] = True             # new key-value pair
```

## Function

```
def kmh_to_ms(speed_kmh):
    """Converts km/h to m/s."""       # docstring: short description
    return speed_kmh / 3.6

kmh_to_ms(95)                          # call with one argument

def kinetic_energy(v, mass=1200):     # mass has a default value
    return 0.5 * mass * v ** 2

kinetic_energy(27.8)                   # mass = 1200
kinetic_energy(27.8, mass=1500)        # mass overridden
```

`def` defines the function, `return` returns a value. Without `return` the
function yields `None`.
